## Your first quantum circuit

Two qubits: a Hadamard gate puts the first one in superposition, and a CNOT entangles the second with it. Then both are measured.

````python
from qiskit import QuantumCircuit

qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0, 1)
qc.measure_all()
qc.draw()
````

Run it 1,000 times on a simulator inside this session (no account needed):

````python
from qiskit.primitives import StatevectorSampler

result = StatevectorSampler().run([qc], shots=1000).result()
print(result[0].data.meas.get_counts())
````

You should see about half `00` and half `11`, and never `01` or `10`: the two qubits are entangled, so measuring one tells you the other.
